```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv('Income.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df['Designation'].unique()
```

```python
for desig in df['Designation'].unique():
    mask=df['Designation']==desig
    plt.scatter(df.Age[mask],df.Income[mask], label=desig)
plt.xlabel('Age')
plt.ylabel('Income')
plt.legend()
plt.show()
```

```python
for desig in df['Designation'].unique():
    mask=df['Designation']==desig
    plt.scatter(df.Age[mask],df.Income[mask], label=desig)
p_age=30
p_income=115000
plt.scatter(p_age,p_income,color='red',label="New Employee")
plt.scatter(35,130000,color='green')
plt.xlabel('Age')
plt.ylabel('Income')
plt.title("Age VS Income for different domains")
plt.legend()
plt.show()
```

```python
model=KNeighborsClassifier(n_neighbors=3)
model
```

```python
x=df.drop(columns=['Designation','Name'])
y=df['Designation']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
model.fit(x_train,y_train)
```

```python
y_predict=model.predict([[32,115000],[35,130000]])

y_predict
```
