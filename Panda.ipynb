 # Create a dataset of your class 
a) Find the best student on basis of marks.
b) Find maximum marks in data analytics and print the name of student.
c) Replace the null value with mean.
d) Find the average of your class in DAA.
e) Print first five rows of your dataset.
f) Print last 5 columns of the dataset.


```python
import pandas as pd
import numpy as np
data = {
    "Roll_No" : [1,19,7,4,69,73,35,49,55,52],
    "Name": ["Aadya", "Aditi", "Ayushi", "Aastha", "Anshika", "Anushka", "Ananya", "Akshita", "Anjil", "Amrita"],
    "DAA": [85, 92, 74, 90, 88, 95, 79, 84, 91, 87],
    "D_A": [78, 95, 82, 90,np.nan, 87, 76, 89, 94, 88],
    "ML": [89, 94, 88, 90, 84, 92, 80, 86, 93, 85],
    "PE": [82, 90, 79, 88, 91,np.nan, 77, 85, 92, 89],
    "CN": [84, 91, 76, 89, 93, 96, 78, 87, 95, 90]
}
```

```python
st_df=pd.DataFrame(data)
st_df
```

```python
st_df.columns
```

```python
# (c) Replace null values with mean
st_df['D_A'] = st_df['D_A'].fillna(st_df['D_A'].mean())
st_df['PE'] = st_df['PE'].fillna(st_df['PE'].mean())
st_df.head()
```

```python
# (d) Find average marks in DAA
daa_mean=st_df["DAA"].mean()
daa_mean
```

```python
st_df["Total"] = st_df[['DAA', 'D_A', 'ML', 'PE', 'CN']].sum(axis=1)
st_df.head()
```

```python
# (a) Find the best student on basis of total marks
best_student = st_df.loc[st_df["Total"].idxmax(), "Name"]
best_student
```

```python
# (b) Find maximum marks in Data Analytics and print student name
max_da = st_df["D_A"].max()
student_da = st_df.loc[st_df["D_A"].idxmax(), "Name"]
max_da, student_da
```
