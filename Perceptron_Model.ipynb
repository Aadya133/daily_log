# Perceptron Model

```python
from sklearn.linear_model import Perceptron
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
import pandas as pd
```

```python
df=pd.read_csv("ieee.csv")
X=df.drop("target",axis=1).values
Y=df["target"].values
print(X.shape,Y.shape)
```

```python
X_train,X_test,Y_train, Y_test=train_test_split(X,Y,test_size=0.3, random_state=42)
print(X_train.shape, X_test.shape)
print(Y_train.shape, Y_test.shape)
```

```python
clf=Perceptron()
```

```python
clf.fit(X_train, Y_train)
```

```python
y_pred=clf.predict(X_test)
print(y_pred)
```

```python
accuracy=accuracy_score(Y_test,y_pred)
print(accuracy)
```

```python
conf_matrix=confusion_matrix(Y_test,y_pred)
cm_display=metrics.ConfusionMatrixDisplay(confusion_matrix=conf_matrix, display_labels=[False,True])
cm_display.plot()
```

# Neural Network

```python
# 3 input - 1 output
inputs=[1.2,5.1, 2.1]
weight=[3.1, 2.1, 8.7]
bias=3.0

output= inputs[0]*weight[0] + inputs[1]*weight[1]+inputs[2]*weight[2] +bias
print(output)
```

```python
import numpy as np
def sigmoid(x):
    return 1/ (1+np.exp(-x))
print(sigmoid(output))
```

```python
# 4 inputs - 3 outputs
input=[1.0, 2.0, 3.0, 2.5]
wt1=[0.2,0.8,-0.5,1.0]
wt2=[0.5,-0.91,0.26,-0.5]
wt3=[-0.26,-0.27,0.17,0.87]
b1=2.0
b2=3.0
b3=0.5

output=[
    input[0]*wt1[0]+input[1]*wt1[1] + input[2]*wt1[2] + input[3]*wt1[3]+ b1,
    input[0]*wt2[0]+input[1]*wt2[1] + input[2]*wt2[2] + input[3]*wt2[3]+ b2,
    input[0]*wt3[0]+input[1]*wt3[1] + input[2]*wt3[2] + input[3]*wt3[3]+ b3
]
print(output)
```

```python
for i in output:
    print(sigmoid(i))
```
