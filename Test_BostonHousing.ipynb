# Linear Regression - Predicting MEDV

```python
import pandas as pd

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Loading the dataset
df=pd.read_csv("BostonHousing.csv")
df.head()
```

```python
# Checking Dataset - Linear Regression (Numerical)
df.info()
```

```python
# Shape of the dataset
df.shape
```

```python
# Checking for Null Values
df.isnull().sum()
```

```python
# Replacing NULL values with mean()
df['CRIM']=df['CRIM'].fillna(df['CRIM'].mean())
df['ZN']=df['ZN'].fillna(df['ZN'].mean())
df['INDUS']=df['INDUS'].fillna(df['INDUS'].mean())
df['CHAS']=df['CHAS'].fillna(df['CHAS'].mean())
df['AGE']=df['AGE'].fillna(df['AGE'].mean())
df['LSTAT']=df['LSTAT'].fillna(df['LSTAT'].mean())

df.isnull().sum()
```

```python
# Separating Features and Target values
x=df[['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX',
       'PTRATIO', 'B', 'LSTAT']] #Features
y=df['MEDV']  #Target -> predict
```

```python
# test-train data split
X_train, X_test, Y_train, Y_test = train_test_split(x,y,test_size=0.2,random_state=25)
X_train.shape, Y_train.shape, X_test.shape, Y_test.shape
```

```python
# fitting the data
model=LinearRegression()
model.fit(X_train,Y_train)
```

```python
# prediction
y_predict=model.predict(X_test)
```

```python
# mean absolute error
mae=mean_absolute_error(Y_test,y_predict)
mae
```

```python
#Plotting Actual vs Predicted MEDV
 
plt.figure(figsize=(6,6))
sns.scatterplot(x=Y_test, y=y_predict)
plt.xlabel("Actual MEDV (y_test)")
plt.ylabel("Predicted MEDV (y_pred)")
plt.title("Actual vs Predicted MEDV")
plt.plot([Y_test.min(), Y_test.max()],
         [Y_test.min(), Y_test.max()],
         color='orange', linewidth=2)  # 45-degree line
plt.show()
```

```python
# Performane of the model
r2_score(Y_test,y_predict)
```
